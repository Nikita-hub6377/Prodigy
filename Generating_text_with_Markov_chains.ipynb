```python
def generate_text(markov_chain, length=50, n=1):
  current_key = random.choice(list(markov_chain.keys()))
  output = list(current_key)

  for _ in range(length):
    next_words = markov_chain.get(current_key)
    if not next_words:
         break
    next_word = random.choice(next_words)
    output.append(next_word)
    current_key = tuple(output[-n:])

  return ''.join(output)

```

```python
markov_chain =build_markov_chain(text, n=1)
generated_text = generate_text(markov_chain, length=100, n=1)
print(generated_text)
```
